# [Step 2] 시간 축(Timeline) 기준 데이터 분할 (Train / Valid / Test)

```python
# 1. 이전 단계에서 저장한 수집 데이터 로드
raw_file_path = DATA_RAW / "raw_ohlcv_100.csv"
if not raw_file_path.exists():
    raise FileNotFoundError(f"[-] 원천 데이터 파일이 존재하지 않습니다: {raw_file_path}")

df_all = pd.read_csv(raw_file_path)
df_all["date"] = pd.to_datetime(df_all["date"])
```

```python
# 2. 전체 데이터셋에 존재하는 모든 유니크 영업일(Trading Days) 추출 후 정렬
unique_dates = sorted(df_all["date"].unique())
total_days = len(unique_dates)

print(f"[+] 전체 영업일수: {total_days}일")
```

```python
# 3. 2년(24개월) / 3개월 / 3개월 비율(24 : 3 : 3)에 맞게 컷오프 인덱스 계산
train_end_idx = int(total_days * (24 / 30))
valid_end_idx = int(total_days * (27 / 30))

train_dates = unique_dates[:train_end_idx]
valid_dates = unique_dates[train_end_idx:valid_end_idx]
test_dates = unique_dates[valid_end_idx:]

print(f"    - Train(훈련) 기간: {train_dates[0].strftime('%Y-%m-%d')} ~ {train_dates[-1].strftime('%Y-%m-%d')} ({len(train_dates)}일)")
print(f"    - Valid(검증) 기간: {valid_dates[0].strftime('%Y-%m-%d')} ~ {valid_dates[-1].strftime('%Y-%m-%d')} ({len(valid_dates)}일)")
print(f"    - Test(테스트) 기간: {test_dates[0].strftime('%Y-%m-%d')} ~ {test_dates[-1].strftime('%Y-%m-%d')} ({len(test_dates)}일)")
```

```python
# 4. 각 기간 리스트에 일치하는 데이터만 필터링하여 분할
train_df = df_all[df_all["date"].isin(train_dates)].reset_index(drop=True)
valid_df = df_all[df_all["date"].isin(valid_dates)].reset_index(drop=True)
test_df = df_all[df_all["date"].isin(test_dates)].reset_index(drop=True)
```

```python
# 5. 분할된 데이터를 데이터 폴더(data/raw/)에 각각 독립된 파일로 적재
train_df.to_csv(DATA_RAW / "train.csv", index=False, encoding="utf-8-sig")
valid_df.to_csv(DATA_RAW / "valid.csv", index=False, encoding="utf-8-sig")
test_df.to_csv(DATA_RAW / "test.csv", index=False, encoding="utf-8-sig")

print("\n" + "="*50)
print("[+] 데이터 세트 분할 완료!")
print(f"    - Train 세트 규모 : {train_df.shape[0]} 행 (Instance)")
print(f"    - Valid 세트 규모 : {valid_df.shape[0]} 행 (Instance)")
print(f"    - Test 세트 규모  : {test_df.shape[0]} 행 (Instance)")
print("="*50)
```
